# Disparity

```python
SENTIMENT_MAPPINGS = {
    'robust_sentiment': {
        "Very Negative": 0,
        "Negative": 1,
        "Neutral": 2,
        "Positive": 3,
        "Very Positive": 4
    },
    'siebert': {
        "NEGATIVE": 0,
        "POSITIVE": 1
    },
    'bert-base': {
        "1 star" : 0,
        "2 stars": 1,
        "3 stars": 2,
        "4 stars": 3,
        "5 stars": 4
    }
}
```

```python
def compute_sentiment_disparity(df, sentiment_classifier, sentiment_to_num):
    # first we map the categorical data to int values
    df['sentiment_score'] = df[sentiment_classifier].map(sentiment_to_num[sentiment_classifier])
    # then we aggregate by profile id
    profile_sentiment = df.groupby('profile_id')['sentiment_score'].agg('mean')

    pairs = []

    for i in range(1, 529):
        male_score = profile_sentiment.loc[i]
        female_score = profile_sentiment.loc[i + 528]
        diff = female_score - male_score  # positive means female received higher sentiment

        pairs.append({
            'profile_id_male': i,
            'profile_id_female': i + 528,
            'sentiment_male_score': male_score,
            'sentiment_female_score': female_score,
            'sentiment_score_diff': diff
        })

    return pd.DataFrame(pairs)

def sentiment_disparity_analsysis(disparity_df):
    mean_diff = disparity_df['sentiment_score_diff'].mean()
    std_diff = disparity_df['sentiment_score_diff'].std()
    direction_bias = disparity_df['sentiment_score_diff'].gt(0).mean()
    parity = (disparity_df['sentiment_score_diff'] == 0).mean()

    t_stat, p_val_t = ttest_rel(disparity_df['sentiment_female_score'], disparity_df['sentiment_male_score'])
    
    return mean_diff, std_diff, direction_bias, parity, t_stat, p_val_t
```

```python
disparity_df = compute_sentiment_disparity(df_gpt, 'siebert', SENTIMENT_MAPPINGS)
print("Model: gpt | Sentiment classifier: SiEBERT")
print("------------------------------------------")
mean_diff, std_diff, direction_bias, parity, t_stat, p_val_t = sentiment_disparity_analsysis(disparity_df)
print(f"Mean difference: {mean_diff:.3f}")
print(f"Std deviation: {std_diff:.3f}")
print(f"Percentage of pairs where female score > male score: {direction_bias:.2%}")
print(f"Percentage of pairs where female score = male score: {parity:.2%}")
print(f"Paired t-test: t = {t_stat:.3f}, p = {p_val_t:.4f}")
```

```python
disparity_df = compute_sentiment_disparity(df_gpt, 'robust_sentiment', SENTIMENT_MAPPINGS)
print("Model: gpt | Sentiment classifier: Robust Sentiment")
print("------------------------------------------")
mean_diff, std_diff, direction_bias, parity, t_stat, p_val_t = sentiment_disparity_analsysis(disparity_df)
print(f"Mean difference: {mean_diff:.3f}")
print(f"Std deviation: {std_diff:.3f}")
print(f"Percentage of pairs where female score > male score: {direction_bias:.2%}")
print(f"Percentage of pairs where female score = male score: {parity:.2%}")
print(f"Paired t-test: t = {t_stat:.3f}, p = {p_val_t:.4f}")
```
